```python
import spacy
nlp=spacy.load("en_core_web_sm")

with open("news_story.txt","r") as f:
    news_text=f.read()

news_text[:500]
```

```python
doc=nlp(news_text)
noun_tokens=[]
numeral_tokens=[]
for token in doc:
    if token.pos_ == "NOUN":
        noun_tokens.append(token)
    if token.pos_ =='NUM':
        numeral_tokens.append(token)
print(noun_tokens)
print(numeral_tokens)
```

```python
count=doc.count_by(spacy.attrs.POS)
count
```
